```python
# Check the anomaly labels across the dataset

from collections import Counter

anomaly_labels = [
    sample["labels"]["anomaly_present"]
    for sample in dataset["full"]
]

print(Counter(anomaly_labels))
```

```python
# Check the anomaly types

anomaly_types = []

for sample in dataset["full"]:
    if sample["anomalies"]["exists"]:
        anomaly_types.append(sample["anomalies"]["type"])

print(Counter(anomaly_types))
```

```python
# Check congestion labels

congestion_labels = [
    sample["labels"]["congestion"]
    for sample in dataset["full"]
]

print(Counter(congestion_labels))
```

```python
# Find the first anomalous sample
for i, s in enumerate(dataset["full"]):
    if s["anomalies"]["exists"]:
        anomaly_sample = s
        anomaly_index = i
        break

print("Anomalous sample index:", anomaly_index)

print("\nLabels:")
print(anomaly_sample["labels"])

print("\nAnomaly information:")
print(anomaly_sample["anomalies"])
```

```python
total_samples = len(dataset["full"])

missing_samples = 0

for i, sample in enumerate(dataset["full"]):
    has_missing = False

    for kpi_name, values in sample["KPIs"].items():
        if any(pd.isna(value) for value in values):
            has_missing = True
            break

    if has_missing:
        missing_samples += 1

print("Total samples:", total_samples)
print("Samples containing missing KPI values:", missing_samples)
```
