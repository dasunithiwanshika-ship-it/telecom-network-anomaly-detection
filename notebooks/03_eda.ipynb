```python
import numpy as np

normal_indices = []
anomaly_indices = []

for i in range(len(data)):
    label = data[i]["labels"]["anomaly_present"]

    if label == "No" and len(normal_indices) < 100:
        normal_indices.append(i)

    elif label == "Yes" and len(anomaly_indices) < 100:
        anomaly_indices.append(i)

    if len(normal_indices) == 100 and len(anomaly_indices) == 100:
        break

print("Normal samples:", len(normal_indices))
print("Anomaly samples:", len(anomaly_indices))
```

```python
normal_features = []
anomaly_features = []

for i in normal_indices:
    df = pd.DataFrame(data[i]["KPIs"])

    normal_features.append({
        "RSRP": df["RSRP"].mean(),
        "DL_BLER": df["DL_BLER"].mean(),
        "UL_BLER": df["UL_BLER"].mean(),
        "TX_Bytes": df["TX_Bytes"].mean(),
        "RX_Bytes": df["RX_Bytes"].mean()
    })

for i in anomaly_indices:
    df = pd.DataFrame(data[i]["KPIs"])

    anomaly_features.append({
        "RSRP": df["RSRP"].mean(),
        "DL_BLER": df["DL_BLER"].mean(),
        "UL_BLER": df["UL_BLER"].mean(),
        "TX_Bytes": df["TX_Bytes"].mean(),
        "RX_Bytes": df["RX_Bytes"].mean()
    })

normal_features_df = pd.DataFrame(normal_features)
anomaly_features_df = pd.DataFrame(anomaly_features)

print("Normal features:")
display(normal_features_df.head())

print("\nAnomaly features:")
display(anomaly_features_df.head())
```

```python
comparison = pd.DataFrame({
    "Normal Mean": normal_features_df.mean(),
    "Anomaly Mean": anomaly_features_df.mean()
})

comparison
```

```python
from collections import Counter

anomaly_type_counts = Counter()

for i in range(len(data)):
    anomaly_info = data[i]["anomalies"]

    if anomaly_info["exists"]:
        anomaly_type = anomaly_info["type"]
        anomaly_type_counts[anomaly_type] += 1

print("Number of anomaly types:", len(anomaly_type_counts))

print("\nAnomaly Type Counts:")
for anomaly_type, count in anomaly_type_counts.most_common():
    print(f"{anomaly_type}: {count}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

anomaly_type_df = pd.DataFrame(
    anomaly_type_counts.items(),
    columns=["Anomaly Type", "Count"]
)

anomaly_type_df = anomaly_type_df.sort_values(
    "Count",
    ascending=False
)

plt.figure(figsize=(12, 6))

plt.bar(
    anomaly_type_df["Anomaly Type"],
    anomaly_type_df["Count"]
)

plt.xlabel("Anomaly Type")
plt.ylabel("Number of Samples")
plt.title("Distribution of Anomaly Types")

plt.xticks(rotation=75)
plt.tight_layout()
plt.show()
```

```python
from collections import Counter

affected_kpi_counts = Counter()

for i in range(len(data)):
    anomaly_info = data[i]["anomalies"]

    if anomaly_info["exists"]:
        for kpi in anomaly_info["affected_kpis"]:
            if kpi != "":
                affected_kpi_counts[kpi] += 1

print("Affected KPI counts:")

for kpi, count in affected_kpi_counts.most_common():
    print(f"{kpi}: {count}")
```

```python
jamming_kpi_counts = Counter()

for i in range(len(data)):
    anomaly_info = data[i]["anomalies"]

    if anomaly_info["exists"] and anomaly_info["type"] == "Jamming":
        for kpi in anomaly_info["affected_kpis"]:
            if kpi != "":
                jamming_kpi_counts[kpi] += 1

print("Jamming samples:", anomaly_type_counts["Jamming"])

print("\nAffected KPIs in Jamming:")
for kpi, count in jamming_kpi_counts.most_common():
    print(f"{kpi}: {count}")
```

```python
antenna_kpi_counts = Counter()

for i in range(len(data)):
    anomaly_info = data[i]["anomalies"]

    if anomaly_info["exists"] and anomaly_info["type"] == "Antenna Failure":
        for kpi in anomaly_info["affected_kpis"]:
            if kpi != "":
                antenna_kpi_counts[kpi] += 1

print("Antenna Failure samples:", anomaly_type_counts["Antenna Failure"])

print("\nAffected KPIs in Antenna Failure:")
for kpi, count in antenna_kpi_counts.most_common():
    print(f"{kpi}: {count}")
```

```python
anomaly_kpi_summary = []

for anomaly_type in anomaly_type_counts:
    kpi_set = set()

    for i in range(len(data)):
        anomaly_info = data[i]["anomalies"]

        if anomaly_info["exists"] and anomaly_info["type"] == anomaly_type:
            for kpi in anomaly_info["affected_kpis"]:
                if kpi != "":
                    kpi_set.add(kpi)

    anomaly_kpi_summary.append({
        "Anomaly Type": anomaly_type,
        "Number of Samples": anomaly_type_counts[anomaly_type],
        "Number of Affected KPIs": len(kpi_set)
    })

anomaly_kpi_summary_df = pd.DataFrame(anomaly_kpi_summary)

anomaly_kpi_summary_df = anomaly_kpi_summary_df.sort_values(
    "Number of Affected KPIs",
    ascending=False
)

anomaly_kpi_summary_df
```
