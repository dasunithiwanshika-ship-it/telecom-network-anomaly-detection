# Day 4 - Feature Engineering for TelecomTS

```python
from datasets import load_dataset

dataset = load_dataset(
    "AliMaatouk/TelecomTS",
    data_files={"full": "**/chunked.jsonl"}
)

data = dataset["full"]

print("Total samples:", len(data))
```

```python
#inspect one sample
sample = data[0]

print(sample.keys())
```

```python
# Identify numerical KPIs
numeric_kpis = [
    "RSRP",
    "DL_BLER",
    "DL_MCS",
    "UL_BLER",
    "UL_MCS",
    "UL_NPRB",
    "UL_SNR",
    "TX_Bytes",
    "RX_Bytes",
    "Estimated_UL_Buffer",
    "PRBs_DL_Current",
    "PRBs_UL_Current",
    "PRB_Utilization_DL",
    "PRB_Utilization_UL",
    "UL_NumberOfPackets",
    "DL_NumberOfPackets"
]

print("Number of numeric KPIs:", len(numeric_kpis))
```

```python
#Create the Feature Extraction Function

import numpy as np

def extract_features(sample):

    features = {}

    for kpi in numeric_kpis:

        values = np.array(sample["KPIs"][kpi])

        features[f"{kpi}_mean"] = values.mean()
        features[f"{kpi}_std"] = values.std()
        features[f"{kpi}_min"] = values.min()
        features[f"{kpi}_max"] = values.max()

    return features
```

```python
features = extract_features(data[0])

print("Number of features:", len(features))

list(features.items())[:10]
```

```python
all_features = []

for i in range(len(data)):
    sample_features = extract_features(data[i])
    all_features.append(sample_features)

print("Total samples processed:", len(all_features))
print("Features per sample:", len(all_features[0]))
```

```python
import pandas as pd

features_df = pd.DataFrame(all_features)

print("Shape:", features_df.shape)

features_df.head()
```

```python
print(features_df.columns.tolist())
```

```python
target = []

for i in range(len(data)):

    label = data[i]["labels"]["anomaly_present"]

    if label == "Yes":
        target.append(1)
    else:
        target.append(0)

print("Total labels:", len(target))
print("Normal samples:", target.count(0))
print("Anomalous samples:", target.count(1))
```

```python
#Add the Target to the Feature DataFrame
features_df["target"] = target

print("Dataset shape:", features_df.shape)
```

## Day 4 Summary

Today, the TelecomTS time-series data was transformed into a machine learning-ready feature dataset.

For each of the 16 numerical KPI variables, four statistical features were extracted:

- Mean
- Standard deviation
- Minimum
- Maximum

This produced 64 numerical features for each sample.

The anomaly label was converted into a binary target:

- 0 = Normal
- 1 = Anomaly

The final dataset contains 32,000 samples and 64 input features with one target column. Basic data-quality checks were also performed, including missing-value and duplicate-row checks.

The processed feature dataset was saved as `telecom_features.csv` for use in subsequent machine learning experiments.
